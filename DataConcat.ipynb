```python
kagg["brand"].unique()
```

```python
kagg["model"].unique()
```

```python
kagg["fuel"].unique()
```

```python
kagg["fuel"] = kagg["fuel"].replace("Gas", "Benzine")
kagg["fuel"] = kagg["fuel"].replace("Plug-in Hybrid", "Hybrid")
```

```python
kagg["fuel"].unique()
```

```python
kagg["body_shape"].unique() # 18k+ null
```

```python
kagg["body_shape"] = kagg["body_shape"].replace("Coupe" , "Sports / Coupe" )
kagg["body_shape"] = kagg["body_shape"].replace("Van" , "Van / Bus" )
```

```python
kagg['body_shape'] = kagg.body_shape.replace({'Cabriolet': 'Convertible', 'Station': 'Estate'})
```

```python
kagg["engine"].unique() 
```

```python
kagg = kagg.drop(columns = "engine")
```

```python
kagg["color"].value_counts() # 18k+ null
```

```python
kagg["color"] = kagg["color"].replace("Blue" , "Blue- Navy Blue")
kagg["color"] = kagg["color"].replace("Navy Blue" , "Blue- Navy Blue")
```

```python
print(kagg.columns.tolist())
```

```python
kagg = kagg.rename(columns={
    'brand':      'make',
    'mileage_km': 'kilometers',
    'price_egp':  'price',
    'fuel':       'fuel_type',
    'body_shape': 'body_type',
    'color':      'exterior_color',
})
```

```python
kagg["transmission"].unique()
```

```python
kagg['transmission'] = kagg.transmission.replace(
    {'Dual Clutch': 'Automatic', 'CVT': 'Automatic', 'XTRONIC CVT': 'Automatic'})
```

```python
print(kagg.columns.tolist())
```

```python
kagg["source"].unique()
```

```python
dub["source"] = "dubizzle" # adding this column for future plans as deployment
print(dub.columns.tolist())
```

```python
dub.groupby("condition")["kilometers"].mean()
```

```python
is_new = (kagg.kilometers <= 100) & (kagg.year >= 2025)
kagg['condition'] = is_new.map({True: 'New', False: 'Used'})
kagg.loc[kagg.kilometers.isna(), 'condition'] = pd.NA
```

```python
kagg.condition.value_counts(normalize=True)
```

```python
dub.condition.value_counts(normalize=True)
```

```python
kagg["payment_options"] = np.nan
```

```python
kagg = kagg[~kagg.listing_id.isin(dub.listing_id)]
```

```python
kagg.to_csv("kagg_cleaned.csv", index=False)
dub.to_csv("dub_cleaned.csv" , index=False)
```
